```python
#create model class
import torch
import torch.nn as nn

class Model (nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)
```

```python
#create a dataset
features = torch.rand(10, 5)

#create model
model = Model(features.shape[1])

#call model for forward pass
model(features)
```

```python
#creating another model
#input 5, Hidden 3, Ouput 1
#input to hidden ReLU
#hidden to output Sigmoid

import torch
import torch.nn as nn

class Model_New (nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear1(features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
features = torch.rand(10, 5)

model = Model_New(features.shape[1])

model(features)
```

```python
#creating a model using Sequential pipeline

class Model_New_2(nn.Module):

  def __init__ (self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

```python
features = torch.rand(10, 5)
model = Model_New_2(features.shape[1])
model(features)
```
