```python
import torch

x = torch.tensor(6.7)
y = torch.tensor(0.0)

#weight
w = torch.tensor(1.0)
#bias
b = torch.tensor(0.0)
```

```python
def binary_cross_entropy_loss(prediction, target):
  epsilon = 1e-8
  prediction = torch.clamp(prediction, epsilon, 1-epsilon)
  return -(target * torch.log(prediction) + (1-target)*torch.log(1-prediction))
```

```python
#forward pass
z = w * x + b
#applying sigmoid activation function
y_pred = torch.sigmoid(z)

#calculating loss using Binary Cross Entropy
loss = binary_cross_entropy_loss(y_pred, y)
```

```python
loss
```

```python
#dL/d(y_pred) : Loss w.r.t prediction
d_loss_dy_pred = (y_pred - y)/(y_pred * (1-y_pred))

#d(y_pred)/dz : Prediction w.r.t z (signmoid derivative)
dy_pred_dz = y_pred * (1-y_pred)

#dz/dw and dz/db : z w.r.t w and b
dz_dw = x
dz_db = 1

dL_dw = d_loss_dy_pred * dy_pred_dz * dz_dw
dL_db = d_loss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"Manual Gradient of loss w.r.t weight (dw): {dL_dw}")
print(f"Manual Gradient of loss w.r.t weight (db): {dL_db}")
```
